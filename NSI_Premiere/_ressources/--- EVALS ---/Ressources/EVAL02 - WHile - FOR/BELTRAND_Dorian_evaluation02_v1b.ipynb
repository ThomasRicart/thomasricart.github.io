```python
## Question 1
def fonction1(n):
    ''' 
    Fonction renvoyant la somme des n premiers entiers naturels.
    Vous utiliserez obligatoirement une structure FOR
    Exemple si n = 6: somme = 1 + 2 + 3 + 4 + 5 + 6
    : param (int): un entier strictement supérieur à 0
    : return (int): un entier
    >>> fonction1(6)
    21
    >>> fonction1(9)
    45
    '''
    assert type(n) == int , ' n doit etre un entier'
    assert n > 0 , 'n doit etre strictement positif'

    ### A compléter
    somme = 0
    for i in range(n+1):
        somme += i

    return somme
```

```python
def quedeschiffres(val):
    seulementchiffre = True
    for i in val:
        if i not in "12346789":
            seulementchiffre = False
    return seulementchiffre

## Question 3.
def fonction3(a):
    '''
    Fonction qui renvoit le premier chiffre impair dans une chaine de chiffres
    Si la chaine de chiffre ne comporte que des chiffres pairs, la valeur renvoyée est un booléen False
    Vous etes autorisé à utiliser plusieurs "return"
    : param (str) un mot de n chiffres. Chaque caractère est un chiffre.
    : return (str)
    Exemple: mot = '422681654' renverra '1' car c'est le premier caractère impair
    >>> fonction3('1344')
    '1'
    >>> fonction3('4428261')
    '1'
    >>> fonction3('22222222222')
    False
    '''
    ### A compléter avec les asserts
    assert type(a) == str
    assert quedeschiffres(a)

    ### A compléter avec le programme
    for i in range(len(a)):
        if int(a[i]) % 2:
            return a[i]
    return False
    
```

## Question 6
Pour valider un code d'entrée de 5 chiffres ( ne commençant pas par 0), on utilise la technique suivante:

* On calcule la somme des 4 1er chiffres 
* Si le résultat est égal au 5ème chiffre, le code est valide.
* Sinon le code est invalide
     

**Exemple** :

    99999 ---> 9+9+9+9 = 36 ---> 6+3 =9 qui est égal au 5ème chiffre (code valide)
    10013 ---> 1+0+0+1 = 2  qui n'est pas égal à 3 (code invalide)

1. Ecrire une fonction `somme_digits(valeur)` qui renvoie la somme des chiffres contenus dans la chaine de caractère `valeur` 

2. Ecrire une fonction `code_est_valide(nombre)` qui renvoie True si le code est valide est False sinon. Vous réutiliserez la fonction `somme_digits`



```python
def somme_digits(valeur):
    ''' Effectue la somme des chiffres contenus dans la chaine de caractère valeur
    : param (str) une chaine de caractères contenant des chiffres
    : return (str)
    >>> somme_digits('123')
    '6'
    >>> somme_digits('9999')
    '36'
    '''
    ## A compléter avec le programme
    somme = 0
    for i in valeur:
        somme += int(i)
    somme = str(somme)
    return somme
```

```python
print(somme_digits('9999'))
```

```python
def code_est_valide(nombre):
    ''' Effectue la vérification de la validité du code
    : param(str): un code de 5 chiffres
    : return (bool)
    >>> code_est_valide('99999')
    True
    >>> code_est_valide('10013')
    False
    '''
    ## A compléter avec le programme
    sch = ""
    for i in range(len(nombre)-1):
        sch += nombre[i]
    
    while len(sch) > 1:
        sch = str(somme_digits(sch))
    
    estvalide = False
    if sch == nombre[-1]:
        estvalide = True
    return estvalide
```

```python
print(code_est_valide('99999'))
```
