## Exercice 1:

Créer une fonction `est_liste_entier(liste)` qui renvoie `True` si le paramètre est une liste et ne contient que des entiers

```python
def est_liste_entier(liste):
    '''Renvoie True si le paramètre est une liste d'entiers
    >>> est_liste_entier([1,2,3])
    True
    >>> est_liste_entier([1,'a',3])
    False
    '''
    if type(liste) != list:
        return False
    for elt in liste:
        if type(elt) != int:
            return False
    return True
        
    
```
