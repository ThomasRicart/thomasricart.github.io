<h2 class="alert alert-info">Exercice 4 : copier une liste</h2>

Lorsque l'on veut modifier une donnée, il est souvent utile de travailler sur une copie de cette donnée afin de ne pas altérer son contenu initial.

Toutefois, comme nous l'avons vu dans une précédente activité, on ne peut pas copier une liste en procédant par simple affectation, puisque l'on copie dans ce cas, non pas la liste, mais l'adresse mémoire du premier élément de la liste. Ainsi, la modification d'une liste entraîne la modification de l'autre liste.

1. Le code ci-dessous illustre, à l'aide de Python Tutor, ce phénomène : les deux listes `L1` et `L2` _pointent_ sur le même objet et donc, lorsque l'on modifie `L2`, on modifie également `L1`.

```python
from tutor import tutor

L1 = [5]
L2 = L1
L2[0] = 3

tutor()
```

Pour contourner ce problème, il faut copier _en profondeur_ la liste, c'est-à-dire son contenu. La méthode `copy` permet de réaliser cette copie.

2. Exécuter la cellule ci-dessus et comparer avec le code précédent.

```python
from tutor import tutor

L1 = [5]
L2 = L1.copy()
L2[0] = 3

tutor()
```
