## Algorithme KNN  :
La méthode des k plus proches voisins est une **méthode d’apprentissage supervisé** utilisée entre autres en IA mais aussi pour la reconnaissance de formes.  C'est un des plus simples algorithmes de machines learning.

**Problème de classification :** on dispose d'un ensemble E d'élements et d'un ensemble C de classes 

Chaque élément ei de E est associé à une et une seule classe cj de C.

L'objectif de l'algorithme de classification est de **prédire la classe c d'un élement e.** 

Pour cela on dispose d’une base de données d'apprentissage constituée de N couples d'associations correctes  (ei,cj) . Pour estimer la classe associée à un nouveau élément e, la méthode des k plus proches voisins consiste à prendre en compte (de façon identique) les k échantillons d'apprentissage les plus proches de e, selon une distance à définir (suivant le problème).

***Remarque :** Lorsque le problème consiste à estimer une valeur numérique (sur une moyenne des voisins ou autre) et non une classe, on parle de **problème de regression.***



## Objectif et description de notre problème de classification :

L’objectif de cette activité est de découvrir cet algorithme avec un problème très simple de classification.

Un survol aérien de la banquise a permis d'obtenir une base de donnée des positions des ours et des phoques à un instant donné. 
Une ONG de protection des animaux vous commande un algorithme permettant de prédire **la nature d'un animal errant (classe c de l'ensemble C)** connaissant **sa position(élément e de l'ensemble E).**
Les ours et les phoques ayant tendance à ne pas se mélanger géograpĥiquement, on décide de trouver **les trois plus proches voisins** de cet animal afin d'émettre une hypothèse cohérente sur son espèce. 

Ensemble C : (ours,phoque)

Ensemble E : ((x0,y0),....(xi,yi))
