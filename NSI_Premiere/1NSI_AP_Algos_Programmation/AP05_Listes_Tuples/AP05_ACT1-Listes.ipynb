### **Introduction : Les listes**

    
Une liste est une structure de donnée linéaire qui contient des `éléments` ou `items` de type quelconque. (int, str, tuple, list).
    
Ces éléments sont placés à l'intérieur de crochets et séparés par les des virgules.
    
On peut représenter une liste comme un tableau de ce type. Ici la liste possède 8 éléments d'indice compris entre 0 et 7
    
| Indice i  | 0 | 1 | 2 | 3 | 4 | 5 | 6 | 7 |
| :-------: |:-:|:-:|:-:|:-:|:-:|:-:|:-:|:-:|
| Elément   | 4 | 6 | 3 | 1 | 2 | 7 | 8 | 9 |     
    
    
En python, une liste est de type `list`
    
Les listes sont des variables **mutables**. On  peut donc les modifier une fois qu'elles ont été créées


```python
L = [2, 4, 5]
print(L[0]) # Affiche 2
print(L[1]) # Affiche 4
print(L[2]) # Affiche 5
print(type(L)) # Affiche <class 'list'>
print(len(L)) # Affiche 3
```

3. Regardons à présent à l'aide de Python Tutor ce qui se passe lorsque l'on fait une copie de liste avec la méthode classique de l'affection.

```python
from tutor import tutor

L1 = [5]
L2 = L1
L2[0] = 3

tutor()
```

Lors de l'exécution pas à pas du code, on s'aperçoit que la variable `L1` ne contient pas la liste `[5]`, mais un référencement à cette liste (en fait l'adresse mémoire du premier élément de la liste) qui est symbolisé par une flèche (on parle alors de _pointeur_ et on dit que `L1` _pointe_ vers la liste `[5]`). Ainsi, lors de l'exécution de l'instruction `L2 = L1`, c'est le référencement qui est copié et non la liste. Les deux variables `L1` et `L2` pointent donc aux mêmes endroits comme le montre les deux flèches. Ainsi, toute modification dans l'une des deux variables `L1` ou `L2` affecte les deux variables simultanément.

!!! danger A retenir
Lorsque l'on veut effectuer la copie d'une liste, on n'utilise jamais une affectation.<br>

Il existe plusieurs méthodes pour copier correctement une liste (c'est-à-dire créer une copie indépendante de l'originale).<br>
