## Question 1:

Ecrire une fonction `renvoie_valeur(dico)` qui prend un dictionnaire en paramètre et renvoie grâce à une boucle les différentes valeurs du dictionnaire sous forme de liste

Voici un résultat possible:

```python
>>> renvoie_valeur(dico)
[13, 10, 16]
```

````python
### Question 1 - A compléter ###
dico = {'Maths' : 13 , 'Physique' : 10 , 'NSI' : 16}

def renvoie_valeur(dico):
    '''
    Renvoie les valeurs de dico sous forme de liste
    '''
    liste_valeurs = []
    for valeurs in dico.values():
        liste_valeurs.append(valeurs)
    return liste_valeurs
````

````python
### A exécuter pour vérifier la liste renvoyée
renvoie_valeur(dico)
````

## Question 3:

Ecrire une fonction `calcule_moyenne(dico)` qui prend un dictionnaire en paramètre et renvoie la moyenne de l'élève

Voici le résultat attendu:

```python
>>> calcule_moyenne_simple(dico)
13.0
```

Vous pourrez utiliser les fonctions déjà écrites aux questions 1 et 2 si vous le voulez.

````python
### Question 3 - A compléter ###
dico = {'Maths' : 13 , 'Physique' : 10 , 'NSI' : 16}

def calcule_moyenne_simple(dico):
    '''
    Renvoie la moyenne des notes de dico
    '''
    nb_valeurs = 0
    somme_valeurs = 0
    for valeur in dico.values():
        somme_valeurs = somme_valeurs + valeur
        nb_valeurs = nb_valeurs + 1
    return somme_valeurs / nb_valeurs
````

````python
### A exécuter pour vérifier la moyenne calculée
calcule_moyenne_simple(dico)
````

## Question 4:

Les notes sont pondérées de la manière suivante:

* Maths: coefficient 3
* Physique: coefficient 2
* NSI: coefficient 4

L'objecttif est de calculer la moyenne de l'élève arrondie à deux chiffres après la virgule.

Ecrire une fonction `calcule_moyenne_ponderee(dico,coeff)`qui prend en paramètre 

* `dico` un dictionnaire contenant les matières et les notes 

    ```python
    dico = {'Maths' : 13 , 'Physique' : 10 , 'NSI' : 16}
    ```

* `coeff` un dictionnaire contenant les matières et les coefficients

    ```python
    coeff = {'Maths' : 3 , 'Physique' : 2 , 'NSI' : 4}
    ```
Le résultat attendu est le suivant:

```python
>>> calcule_moyenne_ponderee(dico,coeff):
13.67
```

On rappelle que l'arrondi se fait grâce à la commande:

```python
round(valeur,p) # arrondi valeur avec p chiffres après la virgule
```

````python
### Question 4 - A compléter ###
coeff = {'Maths' : 3 , 'Physique' : 2 , 'NSI' : 4}
dico = {'Maths' : 13 , 'Physique' : 10 , 'NSI' : 16}

def calcule_moyenne_ponderee(dico,coeff):
    '''
    Renvoie la moyenne des notes de dico
    '''
    nb_coeffs = 0
    somme = 0
    for cle in dico.keys():
        nb_coeffs = nb_coeffs + coeff[cle]
        somme = somme + dico[cle]*coeff[cle]
    return(round(somme/nb_coeffs,2))
````

````python
### A exécuter pour vérifier la moyenne calculée
calcule_moyenne_ponderee(dico,coeff)
````

# Exercice 2

L'objectif de cet exercice est de compter le nombre de fois où apparait certains caractères dans une chaîne de caractère.




## Question 5

On considère la chaîne de caractère suivante:

```python
mot = '2456376457860986'
```

Ecrire une fonction `compte_caractère` qui prend comme paramètre une chaine de caractère et renvoie un dictionnaire dont les clés sont les caractères au format `str` et les valeurs le nombre de fois où ce caractère apparait.

Voici un des résultats possibles
```python
>>> compte_caractere(mot)
{'0':1 , '2':1 , '3':1 , '4':2 , '5':2 , '6':4 , '7':2 , '8':2 , '9':1}
```

````python
### Question 5 - A compléter ###
mot = '2456376457860986'

def compte_caractere(mot):
    dico = {}
    for lettre in mot:
        if lettre in dico.keys():
            dico[lettre] = dico[lettre] + 1
        else:
            dico[lettre] = 1
    return dico
````

````python
### A exécuter pour vérifier la fonction
compte_caractere(mot)
````

## Question 6

Ecrire une fonction `compte_caractere_2_mots(mot1,mot2)` qui renvoie un dictionnaire contenant le nombre de caractères de chaque type présents dans `mot1` et `mot2`.

Vous pouvez si vous le voulez réutiliser les fonctions précédentes

```python
mot1 = '35625'
mot2 = '13664'
```

Un résultat possible est:

```python
>>> compte_caractere_2_mots(mot1,mot2)
{'1':1 , '2':1 , '3':2 , '4':1 , '5':2 , '6':3}
```


````python
### Question 6 - A compléter ###
mot1 = '35625'
mot2 = '13664'

def compte_caractere_2_mots(mot1,mot2):
    dico = {}
    dico_mot1 = compte_caractere(mot1)
    dico_mot2 = compte_caractere(mot2)
    for cle1 in dico_mot1.keys():
        if cle1 in dico:
            dico[cle1] = dico[cle1] + dico_mot1[cle1]
        else:
            dico[cle1] = dico_mot1[cle1]
    for cle2 in dico_mot2.keys():
        if cle2 in dico:
            dico[cle2] = dico[cle2] + dico_mot2[cle2]
        else:
            dico[cle2] = dico_mot2[cle2]
    return dico
````

````python
### A exécuter pour vérifier la fonction
compte_caractere_2_mots(mot1,mot2)
````

Ecrire une fonction `recherche_max(notes,matière)` qui prend comme paramètre une liste d'élèves avec leurs moyennes et une matière de type chaîne de caractère et renvoie le nom de l'élèves qui a la meilleure moyenne dans la matière demandée.

On suppose que les notes sont comprises entre 0 et 20

Voici un exemple de ce que doit faire la fonction

```python
>>> notes = [{'Eleve':'Adam' , 'Maths':10 , 'NSI':18},
        {'Eleve':'Bob' , 'Maths':12 , 'NSI':16},
        {'Eleve':'Chris' , 'Maths':11 , 'NSI':14}]
>>> recherche_max(notes,'Maths')

```

````python
### Question 7 - A compléter ###
notes = [{'Eleve':'Adam' , 'Maths':10 , 'NSI':18},
        {'Eleve':'Bob' , 'Maths':12 , 'NSI':16},
        {'Eleve':'Chris' , 'Maths':11 , 'NSI':14}]

def recherche_max(notes,matiere):
    note_max = 0
    for eleve in notes:
        if eleve[matiere] > note_max:
            note_max = eleve[matiere]
            eleve_max = eleve['Eleve']
    return eleve_max
            
````

````python
### A exécuter pour vérifier la fonction
recherche_max(notes,'Maths')
````
