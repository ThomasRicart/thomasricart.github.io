## 0. Quelques précisions

En python, il est possible d'afficher:
* du texte (chaines de caractères)
* des valeurs de variables



````python
print('Bonjour')        # Affiche le message 'Bonjour' à l'écran
mot = 'Hello'           # Affecte la chaîne de caractères 'Hello' à la variable mot
print(mot)              # Affiche le contenu de la variable mot à l'écran
valeur = 6              # Affecte la valeur entière 6 à la variable valeur
print(valeur)           # Affiche le contenu de la variable valeur à l'écran    
````

Il est également possible d'afficher une combinaison de chaines de caractères et de valeurs de variables



````python
note = 16                                       # Affecte la valeur entière 16 à la variable note
nom = 'Alice'                                   # Affecte la chaîne de caractères 'Alice' à la variable nom     
print(nom, 'a eu la note de ', note, '/20')     # Affiche une combinaison de la variable nom, d'une chaîne de caractères et de la variable note à l'écran
````

## 7. Premières notions de fonctions

Une `fonction` en Python est une "boite" qui possède:
* un nom
* des paramètres
* une sortie




Par exemple, voici la fonction `carre` qui prend pour paramètre une valeur `x` et renvoie `x**2`

```python
def carre(x):
    ''' Renvoie la valeur de x au carré '''
    y = x ** 2
    return y
```

Une fois que cette fonction est crée, il est possible de s'en servir ainsi:

```python
valeur = carre(6)
print(valeur)
```
