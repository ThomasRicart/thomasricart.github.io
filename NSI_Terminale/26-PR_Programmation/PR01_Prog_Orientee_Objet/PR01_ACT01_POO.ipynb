<h2 style="color: red; font-style: 16px italic verdana;">1. En python, tout est objet </h2>

En python, toutes les variables que vous manipulez sont de type différents mais elles ont un point en commun: il est possible de les imaginer comme des objets sur lesquels nous allons avoir des interactions.

Chacun de ces objet peut avoir des caractéristiques: (ex pour une liste: son contenu, sa longueur) que nous allons pouvoir modifier, récupérer etc.

Ces objets pourront même interagir les uns avec les autres.

Nous allons découvrir dans ce chapitre que tout ce que vous manipulez en python est un objet et que nous allons pouvoir nous aussi créer des objets

**Exemple** Comme dit précédemment, vous manipulez déjà des objets. Le premier est la type `list`

````python
L = [2, 4, 7, 4]
L.append(5)
print(L)
````

On remarque la structure un peu différente de la ligne 2 qui ne comporte pas de signe `=`:
```python
L.append(5)
```

On peut imaginer cette commande comme la phrase " Faire subir à `L` qui est de type `list` une opération (appelée `méthode`) nommée `append` qui prend un argument." On dira que `L` est un `objet` ou une `instance` de la classe `list`.

Python ira vérifier si dans les méthodes du type `list` il existe une méthode nommée `append` et l'appliquera.



On aurait pu écrire:

```python
L = L + [5]
```



Nous avons dit que tout est objet et donc que toute opération que l'on utilise est une méthode. Il est donc possible de lister toutes les `methodes` utilisable par un type de variable.

La fonctionnalité suivante permet de faire cette liste pour le type `list`. On reconnait quelques méthodes déjà utilisées. En particulier `append` et `sort`

````python
dir(list)
````

En faisant la même chose avec le type `int`, on obtient le résultat suivant:

```python
['__abs__', '__add__', '__and__', '__bool__', '__ceil__', '__class__', '__delattr__', '__dir__', '__divmod__', '__doc__', '__eq__', '__float__', '__floor__', '__floordiv__', '__format__', '__ge__',
 '__getattribute__', '__getnewargs__', '__gt__', '__hash__', '__index__', '__init__', '__init_subclass__', '__int__', '__invert__',..., 'from_bytes', 'imag', 'numerator', 'real', 'to_bytes']
```

On remarquera:
* que certaines méthodes possèdent des `_` et d'autre non
* qu'il n'y a pas les fonctionnalités de base du type `+` ou `//`

Lorsque les méthodes possèdent un double `_` au début et à la fin, cela veut dire que ce sont des méthodes spéciales. Tester les codes suivants:

````python
a = 5
b = 6
assert type(a) == type(b) == int
c = a + b
d = a.__add__(b)
print(c, d)
````

L'opérateur `+` est donc implémenter dans la classe `int` par la méthode `__add__`. De même:

* `abs(a)` est implémenté par `a.abs()`
* `a == b` est implémenté par `a.__eq__(b)`
* `print(a)` est implémenté par `a.__str__()`.

En imaginant que nous ayont accès à ces classes de type, nous pourrions créer nos propres classes voire modifier le comportement de classes déjà existantes.



<h2 style="color: red; font-style: 16px italic verdana;">2. Création de la première classe </h2>







<div class="alert alert-info" style="border-left: 5px solid #ac763d;border-radius:2px;"> <b><u>Contenu d'une classe</b></u>

Une classe définira des objets et chaque objet sera caractérisé par:

* son identité: son nom
* ses caractéristiques: les attributs
* son état: les valeurs de ses attributs 
* ses fonctionnalités: les méthodes utilisables

<div class="alert alert-success" style="border-left: 5px solid #3c763d;border-radius:2px;"> <b><u>Exemple 1</b></u>:

On considère l'objet mathématique `Point`. Il peut être défini par:

* son nom: par exemple le point nommé `A`
* ses attributs: abscisse et ordonnée
* son état: les valeurs des abscisses et des ordonnées du point `A`
* ses fonctionnalités: pouvoir afficher les coordonnées du point, déplacer ce point etc.

````python
# Déclaration de la classe Point
class Point:
    # Constructeur de la classe Point
    def __init__(self, abs, ord):
        # Définition de ses attributs
        self.x = abs
        self.y = ord
    
````

Ici nous avons:
* Déclaré une nouvel objet de type `Point`
* Défini la fonction qui sera utilisée lors de la création d'une `instance` de la classe `Point`. Cette fonction a 3 paramètres: `self`, `abs` et `ord`. `self` se réfère à l'objet que nous manipulons.
* défini 2 attributs nommés `x` et `y` qui sont des attributs de l'objet de type `Point`. Voilà pourquoi nous les notons `self.x`

<div class="alert alert-info" style="border-left: 5px solid #ac763d;border-radius:2px;"> <b><u>Prototypage d'une classe</b></u>

De manière générale le minimum pour construire une classe est:
```python
class Nom_de_la_classe:
    def __init__(self, param1, param2):
        self.p1 = param1
        self.p2 = param2
```

Nous pouvons maintenant créer des points, pour cela il suffit de créer une `instance` de la classe:

````python
A = Point(3, 4)
````

* `A` est créé et est maintenant de type `Point`
* Lors de l'appel, la fonction `__init__` est appelée.
* `self` prend la valeur `A`
* `x` prend la valeur 3
* `y` prend la valeur 4

Nous ne pouvons pour le moment faire aucune manipulation sur l'opbjet `A` mis à part:

* connaitre la valeur de son attribut `x` ou `y`
* modifier la valeur de son attribut `x` ou `y`


````python
print(A.x)
print(A.y)
A.x = 7
print(A.x)
````

<div class="alert alert-info" style="border-left: 5px solid #ac763d;border-radius:2px;"> <b><u>Accesseurs et mutateurs</b></u>

Vous ne DEVEZ pas manipuler les attributs de cette manière à l'extérieur de la classe pour des raisons de sécurité. 

A la place noius allons créer des accesseurs (`getters`) et des mutateurs (`setters`)

de manière générale, voici la création de ces méthodes:

```python
def get_attribut(self):
    return self.attribut

def set_attribut(self, nouvelle_valeur):
    self.attribut = nouvelle_valeur
```

````python
# Déclaration de la classe Point
class Point:
    # Constructeur de la classe Point
    def __init__(self, abs, ord):
        # Définition de ses attributs
        self.x = abs
        self.y = ord
        
    def get_abs(self):
        return self.x
    def get_ord(self):
        return self.y
    def get_coord(self):
        return (self.x, self.y)
    
    def set_abs(self, abs):
        self.x = abs
    def set_ord(self, ord):
        self.y = ord
    def set_coord(self, coord):
        self.x, self.y = coord
        
````

````python
A = Point(5, 7)
A.set_abs(9)
print(A.get_abs())
A.set_coord((3, 8))
````

<div class="alert alert-info" style="border-left: 5px solid #ac763d;border-radius:2px;"> <b><u>le paramètre self</b></u>

Le paramètre `self` n'est jamais renseigné puisque celui ci se réfère au nom de l'objet qui a été créé. Lorsque nous manipulons le point `A`, `self` prend l'objet nommé `A`.

Si l'on afficher le point créé, on pourrait imaginer écrire:

```python
print(A)
```

Tester le code suivant. Tout ce qui est dit est que cet objet est de type `Point` et nous avons l'adresse mémoire dans lequel cet objet est déposé.

````python
print(A)
````

Il va donc falloir `surcharger` la fonction `print`, c'est à dire la méthode `__repr__`. Cette méthode devra obligatoirement renvoyer une chaine de caractère.

On peut alors ajouter une méthode à la classe `Point`:

```python
def __str__(self):
    return f'({self.x};{self.y})'
```

et tester la commande `print(A)`
